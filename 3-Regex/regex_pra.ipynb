```python
import re
text = "ID/姓名： 123456789 / Denji"

# 使用正則表達式提取 ID 和姓名
regex = r"ID/姓名： (\d+) \/ (\w+)"

# 使用 search / match 方法進行匹配
match = re.search(regex, text)

print(match)
print(match.group(0))  # 輸出整個匹配的字符串
print(match.group(1))  # 輸出第一個捕獲組（ID）
print(match.group(2))  # 輸出第二個捕獲組（姓名）
```

```python
import re
text = "ID/姓名： 123456789 / Denji"

# 使用正則表達式提取 ID 和姓名
regex = r"ID/姓名： (\d+) \/ (\w+)"

# 使用 search / match 方法進行匹配
data = re.findall(regex, text)
print(data)
for i in data:
    print("ID:", i[0])
    print("姓名:", i[1])
```

```python
import re
text = """
    請用以下方式聯絡：
    住家: (02)2311-3838
    公司: (02)2311-3839#345
    手機: 0912-345-678
"""

# 使用正則表達式提取多種電話號碼格式
regex_home = r'住家: (\(\d+\)\d+-\d{4}$)'
homenum = re.findall(regex_home, text, re.MULTILINE)
print("住家電話:" + homenum[0])

regex_company = r'公司: (\(\d+\)\d+-\d{4})(?:#\d+$)'
companynum = re.findall(regex_company, text, re.MULTILINE)
print("公司電話:" + companynum[0])

regex_phone = r'手機: (\d{4}-\d{3}-\d{3})'
phonenum = re.findall(regex_phone, text, re.MULTILINE)
print("手機號碼:" + phonenum[0])

regex = r'((?:\()?\d+.\d+.\d+.\d+)'
allnum = re.findall(regex, text)
print(allnum)
```
