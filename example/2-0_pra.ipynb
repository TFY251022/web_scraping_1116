```python
import requests
from bs4 import BeautifulSoup as bs
```

```python
# 抓取網頁資料
url = "https://www.ptt.cc/bbs/Gossiping/index.html"

my_cookies = { 
    'over18': '1'
    }
r = requests.get(url, cookies = my_cookies)
soup = bs(r.text, "lxml")

print(soup)
```

```python
# 抓取標題
containers = soup.select("div.r-ent div.title a")
print(containers[0])

prefix = 'https://www.ptt.cc'

for ele in containers:
    print(f"標題: {ele.get_text()} 網址: { prefix + ele['href']}")
```
