```python
from selenium import webdriver
from selenium.webdriver.common.by import By

# from webdriver_manager.chrome import ChromeDriverManager



# 設定 Chrome 瀏覽器的選項
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") # Chrome 瀏覽器在啟動時最大化視窗
options.add_argument("--incognito") # 無痕模式
options.add_argument("--disable-popup-blocking") # 停用 Chrome 的彈窗阻擋功能。

# 建立 Chrome 瀏覽器物件
driver = webdriver.Chrome( options=options)
driver.get("https://today.line.me/tw/v2/movie/incinemas/playing")
```

```python
movie_list = driver.find_elements(By.CSS_SELECTOR, ".detailListItem.movieListing-movie")
```

```python
import time

movie_list = driver.find_elements(By.CSS_SELECTOR, ".detailListItem.movieListing-movie")
last_movie = movie_list[-1]
print(f"目前數量:{len(movie_list)}")

while True:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)
    # 取得新的所有元素
    movie_list = driver.find_elements(By.CSS_SELECTOR, ".detailListItem.movieListing-movie")
    if last_movie == movie_list[-1]:
        break

    last_movie = movie_list[-1]
    print(f"目前數量:{len(movie_list)}")
```

```python
all_movie_info = []
for movie in movie_list:
    movie_name = movie.find_element(By.CSS_SELECTOR, ".detailListItem-titles h2").text
    movie_grade = movie.find_element(By.CSS_SELECTOR, ".iconInfo.detailListItem-iconInfo > span.iconInfo-text.text.text--f.text--secondary.text--regular").text
    movie_time = movie.find_element(By.CSS_SELECTOR, ".detailListItem-status > span.text.text--f.text--secondary.text--regular.text--ellipsis1").text
    movie_link = movie.find_element(By.CSS_SELECTOR, ".detailListItem-bookingButton").get_attribute("href")

    all_movie_info.append({
        "電影名稱": movie_name,
        "電影評分": movie_grade,
        "電影時長及上映日": movie_time,
        "電影購票連結": movie_link
    })
print(all_movie_info[0])
```
